```python
n_idr = len(idr_seq)
n_tm = len(tm_seq)

# ------------------------------------------------------------
# Alignment
# ------------------------------------------------------------

with open("alignment.ali", "w") as f:

    # Template:
    # ANXA11 LCD is missing, TM is present in tm.pdb
    f.write(">P1;tm\n")
    f.write(f"structureX:tm:1:A:{n_tm}:A::::\n")
    f.write(tm_seq + "-" * n_idr + "*\n\n")

    # Target:
    # ANXA11(1-185) followed by TM helix
    f.write(">P1;construct\n")
    f.write("sequence:construct:::::::0.00:0.00\n")
    f.write(tm_seq + idr_seq[::-1] + "*\n")


# ------------------------------------------------------------
# Build models
# ------------------------------------------------------------

class MyModel(AutoModel):

    def select_atoms(self):
        # Refine only the ANXA11 LCD.
        # TM begins at residue n_idr + 1 and is left unchanged.
        return Selection(
            self.residue_range("1:A", f"{n_idr}:A")
        )


a = MyModel(
    env,
    alnfile="alignment.ali",
    knowns="tm",
    sequence="construct"
)

# Generate several conformations
a.starting_model = 1
a.ending_model = 10

a.library_schedule = autosched.fast
a.max_var_iterations = 300

# Optional, but I would initially leave this off
a.md_level = refine.fast

a.make()
```
